```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/complaints_realistic_v2_2.csv"
)

print(df.shape)

df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    ngram_range=(1,2)
)

X = vectorizer.fit_transform(
    df["complaint_text"]
)

print(X.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
complaint_1 = [
    "street light not working"
]

complaint_2 = [
    "lamp post has been out for several days"
]

v1 = vectorizer.transform(complaint_1)

v2 = vectorizer.transform(complaint_2)

similarity = cosine_similarity(
    v1,
    v2
)

print(similarity)
```
